```python
# Calculamos R² (qué tan bien el modelo explica los datos históricos)
predicciones_historicas = modelo(anios)
ss_res = np.sum((totales - predicciones_historicas) ** 2)
ss_tot = np.sum((totales - np.mean(totales)) ** 2)
r2 = 1 - (ss_res / ss_tot)
print("R²:", round(r2, 4))
print("Valores predichos por el modelo en años históricos:", predicciones_historicas.round(0))
print("Valores reales:", totales)
```

```python
# Modelo lineal (grado 1)
coef_lineal = np.polyfit(anios, totales, deg=1)
modelo_lineal = np.poly1d(coef_lineal)

# R² del modelo lineal
pred_lineal_historico = modelo_lineal(anios)
ss_res_lineal = np.sum((totales - pred_lineal_historico) ** 2)
r2_lineal = 1 - (ss_res_lineal / ss_tot)

print("=== MODELO LINEAL ===")
print("R²:", round(r2_lineal, 4))
print("Predicciones históricas:", pred_lineal_historico.round(0))
print("Proyección 2030:", round(modelo_lineal(2030)))
print("Proyección 2035:", round(modelo_lineal(2035)))
print("Proyección 2040:", round(modelo_lineal(2040)))
print("Proyección 2048:", round(modelo_lineal(2048)))

print("\n=== MODELO POLINOMIAL (grado 2) ===")
print("R²:", round(r2, 4))
print("Proyección 2030:", round(modelo(2030)))
print("Proyección 2035:", round(modelo(2035)))
print("Proyección 2040:", round(modelo(2040)))
print("Proyección 2048:", round(modelo(2048)))
```
