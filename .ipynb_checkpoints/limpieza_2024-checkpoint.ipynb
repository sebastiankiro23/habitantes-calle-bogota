```python
diccionario_2024 = pd.read_excel(
    "Data.raw/CHC_2024/diccionario_datos_base_anonimizada.ods",
    engine="odf",
    header=1 # la fila 2 (índice 0) tiene los nombres reales de columna
)
diccionario_2024.columns.tolist()
```

```python
# Buscamos filas donde la descripción mencione "razón" o "causa"
filtro = diccionario_2024['Descripción de la variable o columna'].astype(str).str.contains(
    "razón|causa", case=False, na=False
)
diccionario_2024[filtro][['Nombre de la variable o columna', 'Descripción de la variable o columna', 'Dominios (Categorías, valores permitidos)']]
```

```python
# Mostramos el texto completo sin cortar
pd.set_option('display.max_colwidth', None)

print("=== P20 - Causa de inicio ===")
print(diccionario_2024[diccionario_2024['Nombre de la variable o columna'] == 'P20']['Dominios (Categorías, valores permitidos)'].values[0])

print("\n=== P21 - Causa de persistencia ===")
print(diccionario_2024[diccionario_2024['Nombre de la variable o columna'] == 'P21']['Dominios (Categorías, valores permitidos)'].values[0])
```

```python
mapa_causa_inicio_2024 = {
    1: "Conflictos de convivencia o violencia con familiares o personas con las que vivía",
    2: "Un problema de salud físico o mental",
    3: "Pérdida de fuente de ingresos suya o de su hogar",
    4: "Egreso de una institución (de protección o penitenciaria)",
    5: "Consumo de sustancias psicoactivas",
    6: "Por discriminación",
    7: "Abuso sexual",
    8: "Por decisión propia",
    9: "Pérdida de redes de apoyo",
    10: "Amenaza o riesgo para su vida o integridad física",
    11: "Por ser víctima o desplazado del conflicto armado",
    12: "Otra",
}

mapa_causa_persistencia_2024 = {
    1: "Consumo de sustancias psicoactivas",
    2: "Por decisión propia",
    3: "Amenaza o riesgo para su vida o integridad física",
    4: "Influencia de otras personas",
    5: "Falta de trabajo",
    6: "Conflictos o dificultades familiares",
    7: "Abuso sexual",
    8: "Siempre ha sido persona habitante de calle",
    9: "Por ser víctima o desplazado del conflicto armado",
    10: "Por discriminación",
    11: "Ausencia de redes de apoyo adecuadas",
    12: "Otra",
}

df_2024['causa_inicio_texto'] = df_2024['P20'].map(mapa_causa_inicio_2024)
df_2024['causa_persistencia_texto'] = df_2024['P21'].map(mapa_causa_persistencia_2024)

print(df_2024['causa_inicio_texto'].value_counts())
```

```python
df_2024['P10'].value_counts(dropna=False)
```

```python
mapa_sexo_2024 = {1.0: "Hombre", 2.0: "Mujer", 3.0: "Intersexual"}
mapa_si_no_2024 = {1.0: "Sí", 2.0: "No"}

df_2024_limpio = df_2024[[
    "LOCALIDAD", "P12", "P10", "P20", "P21",
    "P30S1", "P30S2", "P30S3", "P30S4", "P30S5", "P30S6", "P30S7", "P30S8"
]].rename(columns={
    "LOCALIDAD": "localidad",
    "P12": "edad",
    "P10": "sexo",
    "P20": "causa_inicio",
    "P21": "causa_persistencia",
    "P30S1": "consume_cigarrillo",
    "P30S2": "consume_basuco",
    "P30S3": "consume_marihuana",
    "P30S4": "consume_alcohol",
    "P30S5": "consume_inhalantes",
    "P30S6": "consume_pepas",
    "P30S7": "consume_cocaina",
    "P30S8": "consume_heroina",
})

df_2024_limpio["sexo"] = df_2024_limpio["sexo"].map(mapa_sexo_2024)
df_2024_limpio["causa_inicio_texto"] = df_2024_limpio["causa_inicio"].map(mapa_causa_inicio_2024)
df_2024_limpio["causa_persistencia_texto"] = df_2024_limpio["causa_persistencia"].map(mapa_causa_persistencia_2024)

for col in ["consume_cigarrillo", "consume_basuco", "consume_marihuana", "consume_alcohol",
            "consume_inhalantes", "consume_pepas", "consume_cocaina", "consume_heroina"]:
    df_2024_limpio[col] = df_2024_limpio[col].map(mapa_si_no_2024)

df_2024_limpio.head()
```
